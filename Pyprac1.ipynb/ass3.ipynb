Problem 1

```python
contacts = {}

while True:
    print("\n1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        name = input("Enter name: ")

        if name in contacts:
            print("Contact already exists")
        else:
            phone = input("Enter phone number: ")
            contacts[name] = phone
            print("Contact added")

    elif choice == "2":
        name = input("Enter name to search: ")

        if name in contacts:
            print(name + ":", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts saved")
        else:
            for name, phone in contacts.items():
                print(name + ":", phone)

    elif choice == "4":
        print("Goodbye")
        break

    else:
        print("Invalid option")
```

Problem 3

```python
marks = {
    "Aman": 82,
    "Priya": 91,
    "Rahul": 45,
    "Sneha": 67,
    "Kiran": 30
}

passed_students = {}
topper_name = ""
topper_marks = -1

for name, mark in marks.items():
    if mark >= 40:
        print(name, "- Pass")
        passed_students[name] = mark
    else:
        print(name, "- Fail")

    if mark > topper_marks:
        topper_marks = mark
        topper_name = name

print("Topper:", topper_name, "with", topper_marks, "marks")
print("Students who passed:", passed_students)
```

Problem 4

```python
riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

mutual_friends = riya_friends & arjun_friends
only_riya = riya_friends - arjun_friends
all_friends = riya_friends | arjun_friends
suggestions_for_riya = arjun_friends - riya_friends

print("Mutual friends:", mutual_friends)
print("Friends only Riya has:", only_riya)
print("Everyone in either list:", all_friends)
print("Friend suggestions for Riya:", suggestions_for_riya)
```

Problem 5

```python
text = input("Enter text: ")
character_counts = {}

for character in text:
    if character != " ":
        if character in character_counts:
            character_counts[character] = character_counts[character] + 1
        else:
            character_counts[character] = 1

print(character_counts)

if len(character_counts) > 0:
    most_frequent_character = ""
    highest_count = 0

    for character, count in character_counts.items():
        if count > highest_count:
            most_frequent_character = character
            highest_count = count

    print("Most frequent character:", most_frequent_character,
          "(" + str(highest_count), "times)")
```

Problem 6

```python
menu = {
    "dosa": 60,
    "idli": 40,
    "vada": 30,
    "coffee": 25,
    "tea": 15
}

total = 0

while True:
    item = input("What would you like to order? ").lower()

    if item == "bill":
        break
    elif item in menu:
        total = total + menu[item]
        print(item, "added to your order")
    else:
        print("Item not available")

if total > 200:
    discount = total * 0.10
    total = total - discount
    print("10% discount applied")

print("Total bill: Rs.", total)
```

Problem 7

```python
cities = [
    ("Hyderabad", 17.38, 78.48),
    ("Mumbai", 19.07, 72.87),
    ("Chennai", 13.08, 80.27),
    ("Delhi", 28.61, 77.20)
]

northern_city = ""
highest_latitude = -90

for city, latitude, longitude in cities:
    print(city, "Latitude:", latitude, "Longitude:", longitude)

    if latitude > highest_latitude:
        highest_latitude = latitude
        northern_city = city

print("Farthest north:", northern_city, "with latitude", highest_latitude)

# Tuples cannot be changed after they are created.
# Trying to assign a new latitude inside a tuple raises a TypeError.
# For example: cities[1][1] = 20.0
```

Problem 8

```python
books = {
    "Python Basics": 3,
    "Data Science 101": 0,
    "Machine Learning": 2,
    "SQL Guide": 1
}

book_name = input("Enter the book name to borrow: ")

if book_name not in books:
    print("Book not found")
elif books[book_name] == 0:
    print("Currently unavailable")
else:
    books[book_name] = books[book_name] - 1
    print("Book issued successfully")

print("Updated books:", books)
```

Problem 9

```python
employees = [
    ("Ravi", "IT"),
    ("Anjali", "HR"),
    ("Suresh", "IT"),
    ("Meena", "Finance"),
    ("Arun", "HR"),
    ("Divya", "IT")
]

departments = {}

for name, department in employees:
    if department not in departments:
        departments[department] = []

    departments[department].append(name)

print(departments)

for department, names in departments.items():
    print(department, "has", len(names), "employees")
```

Problem 10

```python
answer_key = ("B", "C", "A", "D", "B")
wrong_questions = set()
score = 0

for index in range(len(answer_key)):
    answer = input("Enter answer for question " + str(index + 1) + ": ")
    answer = answer.upper()

    if answer == answer_key[index]:
        print("Question", index + 1, "correct")
        score = score + 1
    else:
        print("Question", index + 1, "wrong")
        wrong_questions.add(index + 1)

print("Wrong question numbers:", wrong_questions)
print("Final score:", score, "out of 5")

if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Needs improvement")
```
