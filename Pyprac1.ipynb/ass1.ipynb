Problem 1

```python
age = int(input("Enter your age: "))

if age < 5:
    price = 0
elif age <= 17:
    price = 100
elif age <= 59:
    price = 200
else:
    price = 120

print("Your ticket price is Rs.", price)
```

Problem 2

```python
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5
else:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 8

print("Total bill: Rs.", bill)
```

Problem 4

```python
total_saved = 0
days = 0

while total_saved < 1000:
    savings = int(input("Enter today's savings: "))
    total_saved = total_saved + savings
    days = days + 1

print("Goal reached in", days, "days! Total saved: Rs.", total_saved)
```

problem 5

```python
correct_pin = 4321
attempts = 0
access_granted = False

while attempts < 3 and access_granted == False:
    pin = int(input("Enter PIN: "))
    attempts = attempts + 1

    if pin == correct_pin:
        print("Access granted")
        access_granted = True
    else:
        print("Wrong PIN")

if access_granted == False:
    print("Card blocked")
```

Problem 6

```python
n = int(input("Enter N: "))
even_count = 0
odd_count = 0

for number in range(1, n + 1):
    if number % 2 == 0:
        even_count = even_count + 1
    else:
        odd_count = odd_count + 1

print("Even numbers:", even_count)
print("Odd numbers:", odd_count)
```

problem 7

```python
secret_number = 27
guess_count = 0
guess = 0

while guess != secret_number:
    guess = int(input("Guess the number: "))
    guess_count = guess_count + 1

    if guess > secret_number:
        print("Too high")
    elif guess < secret_number:
        print("Too low")

print("Correct! You took", guess_count, "guesses.")
```

Problem 8

```python
marks = int(input("Enter marks out of 100: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 35:
    print("Grade: D")
else:
    print("Fail")
```

Problem 9

```python
bill_number = int(input("Enter bill number: "))
number = bill_number
digit_sum = 0

while number > 0:
    digit = number % 10
    digit_sum = digit_sum + digit
    number = number // 10

print("Sum of digits:", digit_sum)
if digit_sum > 20:
    print("Congratulations! You win a free gift.")
else:
    print("Sorry, no free gift this time.")
```
