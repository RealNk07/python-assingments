problem 1

```python
marks = [78, 45, 92, 33, 67, 88, 51, 29]
passed = 0

for mark in marks:
    if mark >= 35:
        passed = passed + 1

average = sum(marks) / len(marks)

print("Highest mark:", max(marks))
print("Lowest mark:", min(marks))
print("Class average:", average)
print("Students passed:", passed)
```

problem 2

```python
shopping_list = []

while True:
    item = input("Enter item: ")

    if item.lower() == "done":
        break
    elif item in shopping_list:
        print("Already added")
    else:
        shopping_list.append(item)

print()
for index in range(len(shopping_list)):
    print(str(index + 1) + ".", shopping_list[index])

print("Total items:", len(shopping_list))
```

```python
username = input("Enter username: ")

if len(username) < 5 or len(username) > 12:
    print("Invalid: Username must be 5 to 12 characters long")
elif " " in username:
    print("Invalid: Username cannot contain spaces")
elif username[0].isdigit():
    print("Invalid: Username cannot start with a digit")
else:
    print("Valid username")
```

Problem 4

```python
sentence = input("Enter a sentence: ")
vowels = 0
consonants = 0
spaces = 0

for character in sentence:
    if character == " ":
        spaces = spaces + 1
    elif character.lower() in "aeiou":
        vowels = vowels + 1
    elif character.isalpha():
        consonants = consonants + 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
```

Problem 5

```python
name = input("Enter your name: ")
lowercase_name = name.lower()
backwards_name = ""

for index in range(len(lowercase_name) - 1, -1, -1):
    backwards_name = backwards_name + lowercase_name[index]

if lowercase_name == backwards_name:
    print("Bonus! Your name is a palindrome.")
else:
    print("Your name is not a palindrome.")
```

Problem 6

```python
temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
fahrenheit_temps = []
hottest_index = 0

for index in range(len(temps)):
    if temps[index] > 37:
        print("Heat alert:", days[index])

    if temps[index] > temps[hottest_index]:
        hottest_index = index

    fahrenheit = temps[index] * 9 / 5 + 32
    fahrenheit_temps.append(fahrenheit)

print("Hottest day:", days[hottest_index])
print("Hottest temperature:", temps[hottest_index], "C")
print("Temperatures in Fahrenheit:", fahrenheit_temps)
```

Problem 7

```python
sentence = input("Enter a sentence: ")
words = sentence.split()

if len(words) > 0:
    longest_word = words[0]
    shortest_word = words[0]
    capital_count = 0

    for word in words:
        if len(word) > len(longest_word):
            longest_word = word
        if len(word) < len(shortest_word):
            shortest_word = word
        if word[0].isupper():
            capital_count = capital_count + 1

    print("Longest word:", longest_word)
    print("Shortest word:", shortest_word)
    print("Words starting with capital letter:", capital_count)
else:
    print("Please enter at least one word.")
```

Problem 8

```python
balances = [1500, -200, 3000, -50, 0, 750, -1000]
positive_accounts = []
negative_accounts = []
total_owed = 0

for balance in balances:
    if balance > 0:
        positive_accounts.append(balance)
    elif balance < 0:
        negative_accounts.append(balance)
        total_owed = total_owed + (-balance)

print("Positive accounts:", positive_accounts)
print("Negative accounts:", negative_accounts)
print("Total amount owed:", total_owed)
```

Problem 9

```python
password = input("Enter a password: ")
has_uppercase = False
has_lowercase = False
has_digit = False

for character in password:
    if character.isupper():
        has_uppercase = True
    if character.islower():
        has_lowercase = True
    if character.isdigit():
        has_digit = True

conditions_met = 0
if has_uppercase:
    conditions_met = conditions_met + 1
if has_lowercase:
    conditions_met = conditions_met + 1
if has_digit:
    conditions_met = conditions_met + 1
if len(password) >= 8:
    conditions_met = conditions_met + 1

if conditions_met == 4:
    print("Strong")
elif conditions_met >= 2:
    print("Medium")
else:
    print("Weak")
```

```python
full_name = input("Enter your full name: ")
name_parts = full_name.split()

if len(name_parts) == 0:
    print("Please enter a name.")
else:
    initials = ""
    short_name = ""

    for part in name_parts:
        initials = initials + part[0].upper() + "."

    for index in range(len(name_parts) - 1):
        short_name = short_name + name_parts[index][0].upper() + ". "
    short_name = short_name + name_parts[len(name_parts) - 1].title()

    print("Initials:", initials)
    print("Title case:", full_name.title())
    print("Surname in full:", short_name)
```
